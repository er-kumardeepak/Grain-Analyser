```python

import xml.etree.ElementTree as ET
from pathlib import Path

xml_files = list(Path("/content").rglob("*.xml"))

assert xml_files, "XML file not found!"

xml_path = xml_files[0]
tree = ET.parse(xml_path)
root = tree.getroot()

images = root.findall("image")

labels = root.findall(".//meta/task/labels/label")
if not labels:
    labels = root.findall(".//meta/project/labels/label")

class_names = [
    label.findtext("name")
    for label in labels
]

print("XML File:", xml_path)
print("Classes:", class_names)
print("Total annotated images:", len(images))
print("Total polygons:", sum(
    len(img.findall("polygon"))
    for img in images
))

print("\nFirst 5 image names:")
for img in images[:5]:
    print(img.get("name"))
```
